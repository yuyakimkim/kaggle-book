```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
x = [10, 20, 30, 0, 40, 40]
```

#### rank(pandas)

```python
# pandasのrank関数で順位に変換する
rank = pd.Series(x).rank()
print(rank.values)
```

#### argsort(numpy)

```python
# numpyのargsort関数を2回適用する方法で順位に変換する
order = np.argsort(x)
rank = np.argsort(order)
print(rank)
```
